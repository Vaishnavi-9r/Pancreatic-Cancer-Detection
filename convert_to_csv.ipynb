```python
import os
import random
import pandas as pd

random.seed(42)

image_dir = "Task07_Pancreas/imagesTr"
label_dir = "Task07_Pancreas/labelsTr"

# Ignore hidden files like ._pancreas_001.nii.gz
images = sorted([
    f for f in os.listdir(image_dir)
    if f.endswith(".nii") and not f.startswith("._")
])

print("Total images:", len(images))

pairs = []

for img in images:
    lbl = img

    img_path = os.path.join(image_dir, img)
    lbl_path = os.path.join(label_dir, lbl)

    if os.path.exists(lbl_path):
        pairs.append([img_path, lbl_path])

print("Matched pairs:", len(pairs))

random.shuffle(pairs)

n = len(pairs)

train_end = int(0.7 * n)
val_end = int(0.85 * n)

train = pairs[:train_end]
val = pairs[train_end:val_end]
test = pairs[val_end:]

csv_folder = "csv_lists"
os.makedirs(csv_folder, exist_ok=True)

pd.DataFrame(train, columns=["image", "label"]).to_csv(
    os.path.join(csv_folder, "train.csv"),
    index=False
)

pd.DataFrame(val, columns=["image", "label"]).to_csv(
    os.path.join(csv_folder, "val.csv"),
    index=False
)

pd.DataFrame(test, columns=["image", "label"]).to_csv(
    os.path.join(csv_folder, "test.csv"),
    index=False
)

print("Train:", len(train))
print("Validation:", len(val))
print("Test:", len(test))
print("CSV files created successfully!")
```
